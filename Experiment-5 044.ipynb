```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Discrete time signal from lab slide (no external file needed)
signal = np.array([1, 2, 3, 4, 5, 6, 7, 8])
N = len(signal)
n = np.arange(N)
k = np.arange(N)

# 2. Compute FFT
fft_result = np.fft.fft(signal)
magnitude_spectrum = np.abs(fft_result)
phase_spectrum = np.angle(fft_result)

# 3. Compute IFFT
reconstructed_signal = np.fft.ifft(fft_result)

# 4. Plot outputs
plt.figure(figsize=(10, 8))

plt.subplot(4, 1, 1)
plt.stem(n, signal)
plt.title("Original Discrete Time Signal x[n]")
plt.grid(True)

plt.subplot(4, 1, 2)
plt.stem(k, magnitude_spectrum)
plt.title("Magnitude Spectrum |X[k]|")
plt.grid(True)

plt.subplot(4, 1, 3)
plt.stem(k, phase_spectrum)
plt.title("Phase Spectrum ∠X[k]")
plt.grid(True)

plt.subplot(4, 1, 4)
plt.stem(n, np.real(reconstructed_signal))
plt.title("Reconstructed Signal (IFFT)")
plt.grid(True)

plt.tight_layout()
plt.show()
```
