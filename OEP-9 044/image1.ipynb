```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = r"C:\Users\Dev Radia\Pictures\Screenshots\image1.png"
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Image not found. Check the path!")
else:
    denoised_median = cv2.medianBlur(image, 3)

    sharpening_kernel = np.array([
        [ 0, -1,  0],
        [-1,  5, -1],
        [ 0, -1,  0]
    ], dtype=np.float32)
    sharpened_image = cv2.filter2D(denoised_median, -1, sharpening_kernel)

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(image, cmap="gray")
    plt.title("Original Corrupted Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(denoised_median, cmap="gray")
    plt.title("Smoothed (Median Filter 3x3)")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(sharpened_image, cmap="gray")
    plt.title("Sharpened (Laplacian Filter)")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
