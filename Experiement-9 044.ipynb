```python
import cv2
import numpy as np

# Load image using raw string
image = cv2.imread(r"C:\Users\Dev Radia\Desktop\defender.jpg")

if image is None:
    print("Error: Image not found. Check the path!")
    exit()

# 1. Gaussian Smoothing Filter
kernel_size = (5, 5)
sigma = 1.5
gaussian_kernel = cv2.getGaussianKernel(kernel_size[0], sigma)
gaussian_kernel = np.outer(gaussian_kernel, gaussian_kernel)
smoothed_image = cv2.filter2D(image, -1, gaussian_kernel)

# 2. Sharpening Filter
sharpening_kernel = np.array([
    [-1, -1, -1],
    [-1,  9, -1],
    [-1, -1, -1]
])
sharpened_image = cv2.filter2D(image, -1, sharpening_kernel)

# Standard OpenCV Display (pops up in native GUI windows)
cv2.imshow("Original Image", image)
cv2.imshow("Smoothed Image", smoothed_image)
cv2.imshow("Sharpened Image", sharpened_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# 1. Load the image
image_path = r"C:\Users\Dev Radia\Desktop\defender.jpg"
image = cv2.imread(image_path)

# 2. Pre-filter with Gaussian blur to suppress noise
blurred_image = cv2.GaussianBlur(image, (5, 5), 0)

# 3. Define Laplacian high-pass sharpening kernel
laplacian_kernel = np.array([
    [ 0, -1,  0],
    [-1,  5, -1],
    [ 0, -1,  0]
], dtype=np.float32)

# 4. Apply filter2D convolution
sharpened_image = cv2.filter2D(blurred_image, -1, laplacian_kernel)

# 5. Display in native OpenCV windows
cv2.imshow("Original Image", image)
cv2.imshow("Blurred Image", blurred_image)
cv2.imshow("Sharpened Image (Laplacian)", sharpened_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
